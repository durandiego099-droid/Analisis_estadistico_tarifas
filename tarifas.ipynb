Para calcular las tarifas, descubrí que en # Fusiona los datos de llamadas, minutos, mensajes e Internet con base en user_id y month, debía aplicar el .index.names porque depronto me salían los índices como columnas, lo cual podría ser contraproducente al momento de crear más columnas luego del último .merge, donde uní la información de las tarifas. 

Los siguientes pasos consistieron en aplicar una lógica similar para todos: primero apliqué clip(lower=0) al resultado de la resta de los minutos (mensajes y gb) que un usuario consumió. Lo hice porque si esta resta resultaba en valores negativos, con clip(lower=0) podría convertir a 0 y decir que el usuario en cuestión no superó el límite contratado. Una vez calculada la diferencia entre lo que un cliente usó y el límite que tenía en la tarifa, multipliqué por la columna usd_per_minute (gb y message en su respectivo lugar). Así calculé el cargo extra de cada caso que haya superado el límite contratado. 

Decidí mostrar los resultados en dos líneas. La primera depende de una condición para mostrar a quienes excedieron el límite y tiene cargos extra. La segunda solo es una vista corta con un head() solo con la intención de mostrar el panorama.  

```python
# Calcula el ingreso mensual para cada usuario

diferencia_real= usuarios_xmes_userid['minutes'] - usuarios_xmes_userid['minutes_included']

diferencia_minutos= diferencia_real.clip(lower=0)

usuarios_xmes_userid['cargo_por_minextra']= diferencia_minutos*usuarios_xmes_userid['usd_per_minute']

diferencia_mensajes= usuarios_xmes_userid['messages'] - usuarios_xmes_userid['messages_included']

diferencia_real_mensajes= diferencia_mensajes.clip(lower=0)

usuarios_xmes_userid['cargo_extra_mensajes']= diferencia_real_mensajes*usuarios_xmes_userid['usd_per_message']

diferencia_gb= usuarios_xmes_userid['gb_used'] - usuarios_xmes_userid['gb_per_month_included']

diferencia_real_gb= diferencia_gb.clip(lower=0)

usuarios_xmes_userid['cargo_extra_gb']= diferencia_real_gb*usuarios_xmes_userid['usd_per_gb']

fecha = usuarios_xmes_userid['churn_date'].dt.day
dias = usuarios_xmes_userid['churn_date'].dt.days_in_month
prorrateo = fecha / dias * usuarios_xmes_userid['usd_monthly_pay']
usuarios_xmes_userid['es_mes_de_baja'] = usuarios_xmes_userid['año_y_mes'] == usuarios_xmes_userid['churn_month']
condiciones = [
    usuarios_xmes_userid['activo_en_mes'] == False,
    usuarios_xmes_userid['año_y_mes'] == usuarios_xmes_userid['churn_month']
]

resultados = [
    0,
    prorrateo
]

usuarios_xmes_userid['ingresos_del_mes'] = np.select(
    condiciones,
    resultados,
    default=usuarios_xmes_userid['usd_monthly_pay']
)

usuarios_xmes_userid['ingreso_mensual_total']= usuarios_xmes_userid['cargo_por_minextra']+usuarios_xmes_userid['cargo_extra_mensajes'] + usuarios_xmes_userid['cargo_extra_gb'] + usuarios_xmes_userid['ingresos_del_mes']
print(usuarios_xmes_userid[['cargo_por_minextra', 'cargo_extra_mensajes', 'cargo_extra_gb', 'usd_monthly_pay', 'ingreso_mensual_total']].head())

```

Luego con dt.days_in_month obtuve el total de días que tiene ese mes. Con estos valores hice una división: fecha/dias. Después multipliqué esa fracción (prorrateo) por el pago mensual (usd_monthly_pay).

Una vez preparados los datos, comencé con una serie de tres condiciones: primera condición evaluada: activo_en_mes == False. Si es True, el resultado es 0. Esto captura a los usuarios inactivos ese mes (dados de baja en meses anteriores).
Segunda condición : año_y_mes == churn_month. En este caso, si es True (y no se cataloga en la condición 1), el resultado es prorrateo. Esto captura a los usuarios en su mes exacto de baja y se aplica el prorrateo.
Default (ninguna condición se cumplió): el resultado es usd_monthly_pay completo. Esto captura a los usuarios activos normalmente, los que deben pagar la tarifa normal.
Respecto al np.select, lo hice para evaluar las condiciones en orden, y para cada fila usa el resultado de la primera condición que sea True. Si ninguna es True, usa el default .Al final, obtuve el ingreso real generado por cada usuario ese mes, sea completo o parcial.

## Prueba las hipótesis estadísticas

[Prueba la hipótesis de que son diferentes los ingresos promedio procedentes de los usuarios de los planes de llamada Ultimate y Surf.]


H0: El ingreso promedio de Surf es = al ingreso promedio de Ultimate. H1: El ingreso promedio de Surf es ≠ al ingreso promedio de Ultimate.

Elijo el t-test de muestras independientes porque estoy comparando los ingresos de dos grupos de usuarios distintos e independientes entre sí, es decir, los plans Surf y Ultimate. Elegí equal_var=False porque los dos grupos tiene varianzas diferentes. Respecto a alfa, para esta prueba de hipótesis, elijo un nivel de significancia alfa = 0.05, el valor convencional en análisis estadístico, ya que no existe una razón de negocio específica en este caso que justifique un criterio más estricto o más permisivo

<div class="alert alert-block alert-danger">
<b>Reviewer's comment, Iteration 1</b> <a class="tocSkip"></a>

Tu elección de **alpha** necesita corrección/clarificación. No divides 0.05 entre 2 porque hay dos grupos; para una prueba de dos colas, **alpha** se fija comúnmente en 0.05, a menos que justifiques explícitamente un nivel de significancia diferente. Por favor, indica claramente el **alpha** elegido y la regla de decisión; luego compara el **p-value** con ese **alpha**.

</div>


```python
# Prueba las hipótesis
from scipy import stats as st 

import numpy as np

alpha=0.005

print('varianza de surf:',np.var(surf_fil_ingreso))
print('varianza de ultimate:', np.var(ultimate_fil_ingreso))

results= st.ttest_ind(surf_fil_ingreso, ultimate_fil_ingreso, equal_var=False)

print('valor, p: ', results.pvalue)

if(results.pvalue < alpha):

    print('Rechaza la hipótesis nula')

else: 
    print('No se rechaza la hipótesis nula')
```

Usaré de nuevo el t-test de muestras independientes porque estoy comparando los ingresos de dos grupos de usuarios que se ubican en distintas zonas. Eso significa que el equal_var=False es necesario porque los dos grupos tiene varianzas diferentes. Problema recurrente (alpha): por favor, utiliza un nivel de significancia claramente justificado (comúnmente 0.05 para una prueba de dos colas) y aplica de manera consistente la misma regla de decisión al interpretar el p-value.
